```python
import numpy as np

X = np.arange(1, 11).reshape(-1, 1)
y = 3 * X[:, 0] + 2

print(X)
print(y)
```

```python
from pysr import PySRRegressor
```

```python
model = PySRRegressor(
    niterations=100,
    binary_operators=["+", "-", "*", "/"],
    unary_operators=["sin", "cos", "exp", "log"],
)
```

```python
model.fit(X, y)
```

```python
print(model.equations_)
```

```python
print(model.sympy())
```
